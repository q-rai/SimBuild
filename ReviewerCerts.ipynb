```python
reviewerDF = pd.read_csv('SimBuild2026_reviews.csv')
# posterDF['num'] = posterDF.index

# print(reviewerDF.columns)
# print(reviewerDF['contribution_type'].unique())

def makeCertInfo(reviewer):
    name = '{} {}'.format(reviewer[1]['reviewer_firstname'], reviewer[1]['reviewer_name'])
    if name not in reviews:
        reviews[name] = {'abstract': 0, 'paper': 0}
    else:
        if reviewer[1]['contribution_type'] == 'Extended Abstract':
            reviews[name]['abstract'] += 1
        else:
            reviews[name]['paper'] += 1
        # print(qrPath)
    return reviews 

neededCols = ['contribution_type', 'reviewer_name', 'reviewer_firstname']
reviewerDF = reviewerDF[neededCols]
print(reviewerDF)
reviews = {}

for row in reviewerDF.iterrows():
    makeCertInfo(row)

print(reviews)
```
