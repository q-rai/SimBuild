```python
import pandas as pd

def simplify(long, short):
    if long != '':
        return short
    else:
        return ''

df = pd.read_csv('~/Downloads/simbuild-camera2.csv')

print(df.columns)
df.replace('ok', '', inplace=True)
df.replace('12', '', inplace=True)
df['header'].replace('text', 'header', inplace=True)
df['template'] = df.apply(lambda row: simplify(row['template'], 'template'), axis=1, result_type='expand')
df['auth_affil'] = df.apply(lambda row: simplify(row['auth_affil'], 'auth_affil'), axis=1, result_type='expand')
df['ai'] = df.apply(lambda row: simplify(row['ai'], 'ai'), axis=1, result_type='expand')

df['edits'] = df[['status', 'header', 'auth_affil', 'template', 'ai', 'other', 'email_sent']].T.apply(lambda c: c.str.cat(sep=' '))
df['edits'] = df['edits'].str.strip()
# df['revised paperID']
print(df.edits.unique())

for edit in df['edits'].unique():
    editsDF = df[(df['edits'] == edit) & (df['email_sent'] != 'y')]
    try:
        print(edit, editsDF['revised paperID'].astype(int).to_list())
    except:
        print(edit, editsDF['revised paperID'].to_list())

print(df['template'].value_counts())

templateDF = df[df['edits']=='template']
print(templateDF['revised paperID'].to_list())
# df
```
